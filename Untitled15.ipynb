```python
# EXP-7 Lane Detection

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Step 2: Read the Image
img = cv2.imread('road1.png')

if img is None:
    print("Image not found")
    exit()

# Step 3: Convert to Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Step 4: Display Images
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

plt.show()

# Step 5: Thresholding
threshold = cv2.threshold(
    gray, 150, 255, cv2.THRESH_BINARY
)[1]

plt.imshow(threshold, cmap='gray')
plt.title("Thresholded Image")
plt.axis("off")
plt.show()

# Step 6: Region of Interest (ROI)
height, width = gray.shape

mask = np.zeros_like(gray)

polygon = np.array([[
    (0, height),
    (width, height),
    (width, int(height * 0.6)),
    (0, int(height * 0.6))
]], np.int32)

cv2.fillPoly(mask, polygon, 255)

roi = cv2.bitwise_and(threshold, mask)

plt.imshow(roi, cmap='gray')
plt.title("ROI Masked Image")
plt.axis("off")
plt.show()

# Step 7: Edge Detection (Canny)
edges = cv2.Canny(roi, 50, 150)

plt.imshow(edges, cmap='gray')
plt.title("Edge Detected Image")
plt.axis("off")
plt.show()

# Step 8: Gaussian Blur
blur = cv2.GaussianBlur(edges, (5,5), 0)

plt.imshow(blur, cmap='gray')
plt.title("Smoothed Image")
plt.axis("off")
plt.show()

# Step 9: Detect lines using Hough Transform

lines = cv2.HoughLinesP(
    blur,
    1,
    np.pi/180,
    50,
    minLineLength=50,
    maxLineGap=100
)

line_image = img.copy()

if lines is not None:
    lines = np.array(lines).reshape(-1, 4)

    for x1, y1, x2, y2 in lines:
        cv2.line(
            line_image,
            (int(x1), int(y1)),
            (int(x2), int(y2)),
            (0, 255, 0),
            3
        )

plt.imshow(cv2.cvtColor(line_image, cv2.COLOR_BGR2RGB))
plt.title("Detected Lane Lines")
plt.axis("off")
plt.show()

# Step 10: Final Lane Detection Output
plt.figure(figsize=(10,5))
plt.imshow(cv2.cvtColor(line_image, cv2.COLOR_BGR2RGB))
plt.title("Final Lane Detection")
plt.axis("off")
plt.show()
```
